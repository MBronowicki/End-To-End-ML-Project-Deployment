# Student Performance Indicator

### Life cycle of Machine Learning Project
- Understand the Problem Statement
- Data Collection
- Exploratory Data Analysis (EDA)
- Data Pre-processing
- Model Training

## 1) Problem Statement
- This projects understands how the student's performance (test scores) is affected by other factors (variables) such as Gender, Ethnicity, Parental level of education, Lunch and Test preparation course

## 2) Data Collection
- Data Source - [Kaggle](https://www.kaggle.com/datasets/spscientist/students-performance-in-exams?datasetId=74977)
- The data consists of 8 columns and 1000 rows.
